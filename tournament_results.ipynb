# Tournament

##### 'X' vs. 'O'

```python
num_games = 100
results = {'Minimax vs Random': {'X wins': 0, 'O wins': 0, 'Draws': 0},
            'MCTS vs Random': {'X wins': 0, 'O wins': 0, 'Draws': 0},
            'Minimax vs MCTS': {'X wins': 0, 'O wins': 0, 'Draws': 0},
            'MCTS vs Minimax': {'X wins': 0, 'O wins': 0, 'Draws': 0}}
```

### MiniMax vs. Random Agent

```python
for _ in range(num_games):
    result = tournament.play_game(tournament.minimax, tournament.random_agent)
    if result == 1:
        results['Minimax vs Random']['X wins'] += 1
    elif result == -1:
        results['Minimax vs Random']['O wins'] += 1
    else:
        results['Minimax vs Random']['Draws'] += 1
```

### MCTS vs. Random Agent

```python
for _ in range(num_games):
    result = tournament.play_game(tournament.mcts_agent.mcts, tournament.random_agent)
    if result == 1:
        results['MCTS vs Random']['X wins'] += 1
    elif result == -1:
        results['MCTS vs Random']['O wins'] += 1
    else:
        results['MCTS vs Random']['Draws'] += 1
```

### MiniMax vs. MCTS

```python
for _ in range(num_games):
    result = tournament.play_game(tournament.minimax, tournament.mcts_agent.mcts)
    if result == 1:
        results['Minimax vs MCTS']['X wins'] +=1
    elif result == -1:
        results['Minimax vs MCTS']['O wins'] += 1
    else:
        results['Minimax vs MCTS']['Draws'] += 1  
```

### MCTS vs. MiniMax

```python
for _ in range(num_games):
    result = tournament.play_game(tournament.mcts_agent.mcts, tournament.minimax)
    if result == 1:
        results['MCTS vs Minimax']['X wins'] += 1
    elif result == -1:
        results['MCTS vs Minimax']['O wins'] += 1
    else:
        results['MCTS vs Minimax']['Draws'] += 1
```

```python
 # Print results in a  table format
print(f"{'Matchup':20} {'X wins':10} {'O wins':10} {'Draws':10}")
for matchup, stats in results.items():
    print(f"{matchup:20} {stats['X wins']:10} {stats['O wins']:10} {stats['Draws']:10}")
```

# Time Measurement

### Average Times Per Move

```python
print("\nAverage time per move:")
print(f"Minimax: {tournament.measure_time(tournament.minimax):.4f} seconds")
print(f"MCTS: {tournament.measure_time(tournament.mcts_agent.mcts):.4f} seconds")
print(f"Random: {tournament.measure_time(tournament.random_agent):.4f} seconds")
```

### Average Times Per Move for Various MCTS Iterations

```python
avg_100 = tournament.measure_time_mcts(tournament.mcts_agent.mcts, iterations=100)
avg_500 = tournament.measure_time_mcts(tournament.mcts_agent.mcts, iterations=500)
avg_1000 = tournament.measure_time_mcts(tournament.mcts_agent.mcts, iterations=1000)
avg_5000 = tournament.measure_time_mcts(tournament.mcts_agent.mcts, iterations=5000)
avg_10000 = tournament.measure_time_mcts(tournament.mcts_agent.mcts, iterations=10000)
```

```python
iterations = [100, 500, 1000, 5000, 10000]
times = [avg_100, avg_500, avg_1000, avg_5000, avg_10000]

print("\n\nAverage time per move for various MCTS agents:")
print(f"MCTS with 100 iterations: {avg_100:.4f} seconds")
print(f"MCTS with 500 iterations: {avg_500:.4f} seconds")
print(f"MCTS with 1000 iterations: {avg_1000:.4f} seconds")
print(f"MCTS with 5000 iterations: {avg_5000:.4f} seconds")
print(f"MCTS with 10000 iterations: {avg_10000:.4f} seconds")
```
